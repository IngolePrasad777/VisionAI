### Liabraries importing Toolkit


```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

### Data Understanding & Exploration

```python

df = pd.read_csv(r'C:\Users\ketan sonawane\OneDrive\Desktop\VisionAI\ishihara_ml_dataset_10000_rows.csv')
df.head(10)
```

### Data understanding Toolkit


```python
print(df.head())
print(df.info())
print(df["label"].value_counts())
```

### Data Preprocessing Toolkit


```python
feature_columns = [
    "control_fail",
    "red_fail",
    "green_fail",
    "vanishing_seen",
    "total_correct",
    "total_seen"
]

X = df[feature_columns]
y = df["label"]

# encoding the catagorical data
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

joblib.dump(label_encoder, "label_encoder.pkl")

print(y_encoded)

# Train test Split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.2,
    random_state=42,
    stratify=y_encoded
)
```

### Model Training Toolkit


```python
rf_model = RandomForestClassifier(
    n_estimators=500,
    random_state=42
)

rf_model.fit(X_train, y_train)

# Model's prediction
y_pred = rf_model.predict(X_test)
```

### Model's Performance


```python
# estimating accuracy of the model
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)

# confusion matrix 
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:\n", cm)
```

```python
#saved trained model
joblib.dump(rf_model, "cvd_random_forest_model.pkl")
```
